```python
#program to print area of rectangle.. create class AREA. take values of dimensions as parameters of its construcotrs and have a method named returnarea
class area:
  def __init__(self,length,breadth):
    self.len=length
    self.bre=breadth
  def returnarea(self):
    area=self.len*self.bre
    return area

rect=area(3,6)
rect.returnarea()
```

```python
from abc import ABC, abstractmethod
#from abc import
class test(ABC):
  @abstractmethod
  def m1(self):
    pass
  @abstractmethod
  def m2(self):
    pass

class mypython(test):
  def m1(self):
    print("m1 implementation") #an abstract method needs to be defined in the child class

  def m2(self):
    print("m2 implementation")


m=mypython()
m.m1()
m.m2()
```

```python
#100% ABSTRACTION: all functions are abstract in a class
```

If the child class is unable to provide implememntation of all methods of abstract class then we can declare the child class methods with @abstractmethod and complete the remaining method implementations in next created child classes.

```python
from abc import ABC, abstractmethod
#from abc import
class test(ABC):
  @abstractmethod
  def m1(self):
    pass
  @abstractmethod
  def m2(self):
    pass

class mypython(test):
  def m1(self):
    print("m1 implementation") #an abstract method needs to be defined in the child class
  @abstractmethod
  def m2(self):
    pass

class child(mypython):
  def m2(self):
    print("m2 implementation")


m=child()
m.m1()
m.m2()
```

```python
#inside the abstract class we can declare the constructor
#abstract class constructor is executed but object is not created

from abc import ABC, abstractmethod
#from abc import
class test(ABC):
  def __init__(self):
    print("constructor is called")

  @abstractmethod
  def m1(self):
    pass


class mypython(test):
  def m1(self):
    print("m1 implementation")


m=mypython()
m.m1()
```

```python
#wap to print number of lines,words,characters present in a file
import os,sys
fname=input("enter file name:")
if os.path.isfile(fname):
  print("file exists:",fname)
  f=open(fname,'r')
else:
  print("file does not exist:",fname)
  sys.exit(0)

lcount=wcount=ccount=0
for line in f:
  lcount=lcount+1
  ccount=ccount+len(line)
  words=line.split()
  wcount=wcount+len(words)

print("total lines:",lcount)
print("total words:",wcount)
print("total characters:",ccount)
```
